```python
# Step 1: Install dependencies
!pip install crewai openai python-dotenv

# Step 2: Import libraries
from crewai import Agent, Task, Crew
from dotenv import load_dotenv
import os

# Step 3: Load your API key securely
# In Colab, add this manually in the UI: Runtime > Run all, then enter your key when prompted.
os.environ["OPENAI_API_KEY"] = input("Enter your OpenAI API key: ")

# Step 4: Define your agents
career_coach = Agent(
    name="Career Coach",
    role="Career guidance specialist",
    goal="Help users plan and improve their career path.",
    backstory="An experienced HR strategist who helps people identify strengths and set actionable career goals."
)

resume_expert = Agent(
    name="Resume Expert",
    role="Professional resume optimizer",
    goal="Review and rewrite resumes for clarity, impact, and keyword optimization.",
    backstory="A senior recruiter skilled in crafting compelling professional profiles."
)

job_matcher = Agent(
    name="Job Matcher",
    role="Job search strategist",
    goal="Find potential roles that fit a user's background and goals.",
    backstory="An AI agent who knows current job market trends and matching algorithms."
)

# Step 5: Define a collaborative task
career_task = Task(
    description=(
        "Work together to analyze a user's career background, "
        "suggest ideal roles, and recommend improvements to their resume and skills."
    ),
    agents=[career_coach, resume_expert, job_matcher]
)

# Step 6: Create the crew and run it
crew = Crew(agents=[career_coach, resume_expert, job_matcher], tasks=[career_task])
result = crew.run()
print(result)
```
